```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "Qwen/Qwen3-0.6B"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="auto"
)
```

```python
def generate_app(prompt, max_tokens):

    if not prompt.strip():
        return "Please enter a topic or prompt first!"

    messages = [
        {"role": "user", "content": prompt}
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=int(max_tokens),
            do_sample=True,
            temperature=0.7,
            top_p=0.9
        )

    generated_tokens = outputs[0][
        inputs["input_ids"].shape[-1]:
    ]

    generated_text = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    word_count = len(generated_text.split())

    return (
        f"{generated_text}\n\n"
        f"--- Stats ---\n"
        f"Generated words: {word_count}"
    )
```

```python
import gradio as gr

app = gr.Interface(
    fn=generate_app,

    inputs=[
        gr.Textbox(
            label="Enter your topic or prompt",
            placeholder="Write about Artificial Intelligence...",
            lines=6
        ),

        gr.Slider(
            minimum=50,
            maximum=500,
            value=150,
            step=50,
            label="Maximum new tokens"
        )
    ],

    outputs=gr.Textbox(
        label="Generated Text + Stats",
        lines=12
    ),

    title="AI Text Generator — Qwen3",

    description=(
        "Enter a topic or prompt. "
        "The AI generates new text based on your input."
    )
)
```

```python
app.launch(share=True)
```
